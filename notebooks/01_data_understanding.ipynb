# 01 — Data Understanding

This notebook documents the initial exploration of the public IBM Telco Customer Churn dataset and the synthetic operational tables.

## Objectives
- Inspect the schema and data types
- Understand the business context of each field
- Identify potential data quality issues before cleaning
- Document the data provenance

## 1. Load Customer Data

```python
import pandas as pd
import numpy as np
import os

BASE_DIR = os.path.dirname(os.path.abspath("__file__"))
PROCESSED_DIR = os.path.join(BASE_DIR, "..", "data", "processed")

# Load processed customer data (after ingestion and cleaning)
customer_df = pd.read_csv(os.path.join(PROCESSED_DIR, "processed_customer.csv"))

print(f"Shape: {customer_df.shape}")
print(f"\nColumns: {list(customer_df.columns)}")
customer_df.head()
```

## 2. Data Types and Missing Values

```python
print("Data Types:")
print(customer_df.dtypes)
print("\n\nMissing Values:")
print(customer_df.isnull().sum()[customer_df.isnull().sum() > 0])
```

## 3. Summary Statistics

```python
customer_df.describe(include="all").T
```

## 4. Churn Distribution

```python
print(customer_df["churn"].value_counts())
print(f"\nRenewal rate: {(customer_df['churn'] == 'No').mean() * 100:.2f}%")
print(f"Churn rate: {(customer_df['churn'] == 'Yes').mean() * 100:.2f}%")
```

## 5. Synthetic Data Overview

```python
# Load synthetic tables
interactions = pd.read_csv(os.path.join(PROCESSED_DIR, "processed_interactions.csv"))
content = pd.read_csv(os.path.join(PROCESSED_DIR, "processed_content_usage.csv"))
participation = pd.read_csv(os.path.join(PROCESSED_DIR, "processed_event_participation.csv"))
activity = pd.read_csv(os.path.join(PROCESSED_DIR, "processed_account_activity.csv"))

print("Synthetic Tables Overview:")
print(f"- Interactions: {len(interactions)} records")
print(f"- Content Usage: {len(content)} records")
print(f"- Event Participation: {len(participation)} records")
print(f"- Account Activity: {len(activity)} records")

# Unique customers in each table
print(f"\nUnique customers in each table:")
print(f"- Interactions: {interactions['customer_id'].nunique()}")
print(f"- Content Usage: {content['customer_id'].nunique()}")
print(f"- Event Participation: {participation['customer_id'].nunique()}")
print(f"- Account Activity: {activity['customer_id'].nunique()}")
```
